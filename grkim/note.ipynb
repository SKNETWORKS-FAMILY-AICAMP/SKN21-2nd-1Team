- customer_id : 고객 id
- credit_score : 고객 신용점수
- country : 국적
- gender : 성별 (Female/Male)
- age : 나이
- tenure : 은행과 거래한 기간(년 단위)
- balance : 계좌 잔액
- products_number : 고객이 보유한 은행 상품 개수
- credit_card : 신용카드 보유 여부 (0/1)
- active_member : 활성 고객 여부 (0/1)
- estimated_salary : 고객 추정 연봉
- churn : 이탈 여부 (0/1)

```python
# 결측치 확인
df.isnull().sum()
```

```python
category_cols = ['gender', 'country', 'products_number', 'credit_card', 'active_member']

for i in list(range(0, len(category_cols), 2)):
    is_one_col = i + 1 >= len(category_cols)
    left_col = category_cols[i]
    left_data = df[left_col].value_counts()
    
    plt.figure(figsize=(12,5))
    
    print("** ", left_data)

    plt.subplot(1,2,1)
    left_data.plot(kind='bar')
    plt.title(left_col)
    
    if not is_one_col:
        right_col = category_cols[i+1]
        right_data = df[right_col].value_counts()
        print("** ", right_data)

        plt.subplot(1,2,2)
        right_data.plot(kind='bar')
        plt.title(right_col)
    
    # plt.tight_layout()
    plt.show()
```

```python
numeric_cols = ["age", "balance", "tenure", "credit_score", "estimated_salary"]

for col in numeric_cols:
    mean = df.groupby(col)["churn"].mean()
    
    plt.figure(figsize=(12,5))

    plt.subplot(1,2,1)
    sns.kdeplot(df[col], shade=True)
    plt.title(f"{col} Distribution")
    
    plt.subplot(1,2,2)
    plt.bar(mean.index, mean.values)
    plt.title(f"{col} - churn Mean")
    plt.ylim(-1,1)
    
    plt.tight_layout()
    plt.show()
```

```python
df.groupby("tenure")["churn"].mean()
```

```python
corr_list = numeric_cols + ["churn"]
print(corr_list)
corr = df[corr_list].corr()
print(corr)

plt.figure(figsize=(12,12))
sns.heatmap(corr, annot=True)

plt.tight_layout()
plt.show()
```
